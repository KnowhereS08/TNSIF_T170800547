```python
'''
==============================
Q. Data Type Identification Take an input and print its data type (int, float, or string)

==============================
'''
def identify_data_type(user_input):
    # Try to convert to integer
    try:
        val = int(user_input)
        return "int"
    except ValueError:
        pass

    # Try to convert to float
    try:
        val = float(user_input)
        return "float"
    except ValueError:
        pass

    # Default to string
    return "string"

# Take input from user
user_input = input("Enter something: ")
dt = identify_data_type(user_input)
print(f"The data type is: {dt}")

```

```python
'''
 ==============================
 Q. Multi-Class Classification
 If score >= 80 → "A"
 elif >= 60 → "B"
 else → "C"
 ==============================
'''
try:
    score = float(input("Enter your score: "))
    if score >= 80:
        print("Grade: A")
    elif score >= 60:
        print("Grade: B")
    else:
        print("Grade: C")
except ValueError:
    print("Error: Please enter a valid numeric score!")
```

```python
# ==============================
# 📌 Q. Risk Detection
# If age > 60 and bp > 140 → "High Risk"
# else → "Normal"
# ==============================

try:
    age_input = int(input("Enter patient age: "))
    bp_input = int(input("Enter patient systolic blood pressure (BP): "))

    if age_input > 60 and bp_input > 140:
        print("High Risk")
    else:
        print("Normal")
except ValueError:
    print("Error: Please enter valid integers for age and blood pressure!")
```

```python
# ==============================
# 📌 Q. Pass/Fail Prediction
# If marks >= 40 → "Pass"
# else → "Fail"
# ==============================

try:
    marks = float(input("Enter student marks: "))
    if marks >= 40:
        print("Pass")
    else:
        print("Fail")
except ValueError:
    print("Error: Please enter a valid number for marks!")
```

```python
# ==============================
# 📌 Q. Simple Score Prediction
# Take marks of 3 subjects as input
# Print total and average
# ==============================

try:
    subject1 = float(input("Enter marks for Subject 1: "))
    subject2 = float(input("Enter marks for Subject 2: "))
    subject3 = float(input("Enter marks for Subject 3: "))

    total_marks = subject1 + subject2 + subject3
    average_marks = total_marks / 3

    print(f"Total Marks: {total_marks}")
    print(f"Average Marks: {average_marks:.2f}")
except ValueError:
    print("Error: Please enter valid numerical marks!")
```

```python
class Dog:
  species = "Canine" # Class Attribute

  def __init__(self, name, age):
    self.name = name # Instance Attribute
    self.age = age # Instance Attribute

#Creating an object of a Dog Class
dog1 = Dog("Tuffy", 3)
dog2= Dog("Brownie", 7)

print (dog1.name)
print (dog1.age)
print (dog1.species)

print (dog2.name)
print (dog2.age)
print (dog2.species)
```

```python
class Dog:
    species = "Canine"  # Class attribute

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def animal_name(self, dog, cat):
        self.dog = dog
        self.cat = cat
        print("Dog name:", self.dog)
        print("Cat name:", self.cat)
        print("Dog name of init fun ", self.name)

    def location(self, state, country):
        self.state = state
        self.country = country
        print("State:", self.state)
        print("Country:", self.country)



# Taking input from user
name = input("Enter dog object name: ")
age = int(input("Enter age: "))

dog_value = input("Enter dog name: ")
cat_value = input("Enter cat name: ")

# Creating object
d = Dog(name, age)
print (d.name)
print (d.age)
print (d.species)
# Calling function with user input
d.animal_name(dog_value, cat_value)

d.location("Maharashtra", "India")

```
